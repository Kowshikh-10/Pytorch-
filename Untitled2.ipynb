```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class Model(nn.Module):
  def __init__(self, in_features=4, h1=8, h2=9, out_features=3):
    super().__init__()
    self.fc1 = nn.Linear(in_features, h1)
    self.fc2 = nn.Linear(h1, h2)
    self.out = nn.Linear(h2, out_features)

  def forward(self, x):
    x = F.relu(self.fc1(x))
    x = F.relu(self.fc2(x))
    x = self.out(x)

    return x
```

```python
torch.manual_seed(41)
model = Model()
```

Load Data and Train NN


```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
url = 'https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv'
my_df = pd.read_csv(url)
```

```python
my_df
```

```python
#change last col to int
my_df['species'] = my_df['species'].replace('setosa',0.0)
my_df['species'] = my_df['species'].replace('versicolor',1.0)
my_df['species'] = my_df['species'].replace('virginica',2.0)
my_df
```

```python
# Train Test Split! set X, y
X = my_df.drop('species', axis=1)
y = my_df['species']
```

```python
X = X.values
y = y.values
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Train Test split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=41)
```

```python
#convert to float tensors
X_train = torch.FloatTensor(X_train)
X_test = torch.FloatTensor(X_test)
```

```python
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

```python
# set the criterian of model to measure the error , how far off the predictions are from
criterion = nn.CrossEntropyLoss()
# choose Adam optimiser,lr = learning rate
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)

```

```python
import torch

# Train our model
#Epochs?
epoches = 100;
losses = []
for i in range(epoches):
  #go forward
  y_pred = model.forward(X_train)

  #loss/error
  loss = criterion(y_pred,y_train)
  #keep track of our losses
  losses.append(loss.detach().numpy())
  #print every 10 epoch
  if i % 10 == 0:
    print(f'Epoch:{i} and loss: {loss}')

  #Do some backtracking
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
```

```python
# Graph of learning curve
plt.plot(range(epoches),losses)
plt.ylabel('loss')
plt.xlabel('Epoch')
```
